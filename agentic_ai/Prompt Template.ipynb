```python
from databricks_langchain import ChatDatabricks
from langchain_core.prompts import PromptTemplate

llm = ChatDatabricks(model="databricks-gemma-3-12b")

prompt_template=PromptTemplate(
    input_variable=["country","no_of_para","language"],
    template="""You are an expert in traditional cuisines.
    You provide information about a specific dish from a specific country.
    Avoid giving information about fictional places. If its fictional place just give output as 'I dont know'.
    Answer the question: What is the traditional cuisine of {country}.
    Anwer in {no_of_para} paragraphs in {language} language.
    """
)

response=llm.invoke(prompt_template.format(country="India",no_of_para=2,language="Hindi"))
print(response.content)
```

```python
from databricks_langchain import ChatDatabricks
from langchain_core.prompts import PromptTemplate

llm = ChatDatabricks(model="databricks-gemma-3-12b")

prompt_template=PromptTemplate(
    input_variable=["country","no_of_para","language"],
    template="""You are an expert in traditional cuisines.
    You provide information about a specific dish from a specific country.
    Avoid giving information about fictional places. If its fictional place just give output as 'I dont know'.
    Answer the question: What is the traditional cuisine of {country}.
    Anwer in {no_of_para} paragraphs in {language} language.
    """
)


print("Expert in Traditional Cuisine App")

country=input("Enter a country")
no_of_para=input("How many para in number")
language=input("Enter a language")


if country:
    response=llm.invoke(prompt_template.format(country=country,no_of_para=no_of_para,language=language))
    print(response.content)
```
