```sql
create view dev.demo.orderpriority_count as 
select 
o_orderpriority, count(*) as count 
from samples.tpch.orders 
group by o_orderpriority 
order by count desc
```

```sql
create function dev.demo.datamask(x string)
returns string
return concat(repeat("*", length(x)-2),right(x,2)
);
```

```sql
select dev.demo.datamask("naval@g.com")
```

```sql
select *,dev.demo.datamask(email_address) as mask_email  from samples.bakehouse.sales_customers
```
