```python
dbutils.widgets.text("catalog","default")
catalog=dbutils.widgets.get("catalog")
```

```python
df_sales=spark.read.table(f"{catalog}.bronze.sales")
```

```python
df_sales_cleaned=df_sales.dropDuplicates().dropna().drop("file_name","file_path","ingestion_date")
```

```python
df_sales_cleaned.write.mode("overwrite").saveAsTable(f"{catalog}.silver.sales_cleaned")
```
