```python
data=('1','Rohit'),('2',"Virat")
schema="id string, player_name string"

df=spark.createDataFrame(data,schema)
```

```python
df.write.mode("overwrite").saveAsTable("dev.demo.cricket")
```

```python
%sql
select * from dev.demo.cricket
```
