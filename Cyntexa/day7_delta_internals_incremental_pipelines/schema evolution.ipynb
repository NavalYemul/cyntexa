```python
data=(1,'Rohit'),(2,"Virat")
schema="id int, player_name string"

df=spark.createDataFrame(data,schema)
```

```python
df.createOrReplaceTempView("player_view")
```

```python
%sql
create table dev.demo.players as 
select * from player_view
```

```python
df.write.mode("overwrite").saveAsTable("dev.demo.players")
```

```python
%sql
select * from dev.demo.players
```
