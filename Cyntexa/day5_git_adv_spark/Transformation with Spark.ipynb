#withColumnRenamed

```python
df.withColumnRenamed("TransactionID","transaction_id")\
.withColumnRenamed("CustomerID","customer_id")\
.display()
```

```python
df.withColumnRenamed("TransactionID","transaction_id")\
.withColumnRenamed("CustomerID","customer_id")\
.display()
```

```python
df.withColumnsRenamed({"TransactionID":"transaction_id","CustomerID":"customer_id","AccountType":"account_type"})
```

```python
new_col=['transaction_id','customer_id','account_type','transaction_type','amount_usd','branch','transaction_date']
df_renamed_col=df.toDF(*new_col)
```

```python
df_renamed_col.display()
```

```python
from pyspark.sql.functions import *
```

```python
df_renamed_col.withColumn("ingestion_date",current_timestamp())
```

```python
df_renamed_col.withColumn("branch_new",upper(col("branch"))).display()
```
