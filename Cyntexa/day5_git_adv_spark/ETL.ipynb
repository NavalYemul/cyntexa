```python
from pyspark.sql.functions import *
```

```python
df=spark.read.csv("/Volumes/dev/demo/raw/financial_dataset.csv",header=True,inferSchema=True)
```

```python
new_col=['transaction_id','customer_id','account_type','transaction_type','amount_usd','branch','transaction_date']
df1=df.toDF(*new_col)

df_final=df1.withColumn("ingestion_date",current_date()).withColumn("file_name",col("_metadata.file_name"))
```

```python
df_final.write.mode("overwrite").saveAsTable("dev.demo.finance")
```

```python
%sql
create view dev.demo.total_amount as 
select transaction_type, round(sum(amount_usd)) as total_amount from dev.demo.finance group by transaction_type
```

```python
%sql
select * from dev.demo.total_amount
```
